```python
data=load("TE_8_190.sobj") # Dictionary of solutions {n: [primal solution, dual solution, objective function]}
```

```python
N_lower=8 # number of qubits starting from 8
N_upper=190 # number of qubits up to 151

check=0

for n in range(N_lower,N_upper+1):
    
    k=floor(n/2)
    
    x_sol,y_sol,_=data[n] # primal and dual solutions

    A,b,c=Standard_form(n,k) # LP matrices

    if A*x_sol.transpose() >= b and x_sol >= 0 and y_sol*A <= c and y_sol >= 0 and c*x_sol.transpose()==y_sol*b:  
        
        """
        This condition checks if both solutions satisfy the desired conditions + strong duality, i.e.
    
        Primal            Dual
        
        min c*x^T           max b*y
        s.t. Ax >= b      s.t. yB <= c
             x_i >= 0            y_i >= 0
    
        Strong duality: c*x^T = y*b
        
        """
        
        check +=1

check == N_upper-N_lower+1 #  Should be True, meaning that all primal and dual solutions satisfy all conditions + strong duality
```

```python
for n, value in data.items(): # Here we check that all obtained solutions satisfy the conditions to be a (floor(n/2)-1)-TE state
    
    x=value[0][0] # primal solution
    
    for i in range(floor(n/2)):
        if x[i] >= (4/3)*comb(n,i)*2**-i: # APPT condition is satified for each a'_i for a given n
            print(n,i,"error")

# If nothing is printed means that all solutions are compatible with the existence of a (floor(n/2)-1)-TE state for a given n
```
