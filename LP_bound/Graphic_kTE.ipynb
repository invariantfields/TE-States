```julia
N=40 

kuni=[3,
 3,
 3,
 4,
 5,
 5,
 5,
 5,
 5,
 6,
 7,
 7,
 7,
 7,
 7,
 8,
 9,
 9,
 9,
 9,
 9,
 10,
 11,
 11,
 11,
 11,
 11,
 12,
 13,
 13,
 13,
 13,
 13]
kte=[3,
 4,
 4,
 5,
 5,
 6,
 6,
 7,
 7,
 8,
 8,
 9,
 9,
 10,
 10,
 11,
 11,
 12,
 12,
 12,
 13,
 13,
 14,
 14,
 15,
 15,
 16,
 16,
 17,
 17,
 18,
 18,
 19]
kame=[floor(n/2) for n in 8:N]
```

```julia
using Plots, LaTeXStrings

X=8:N
plot(
    X, kame,
    lw = 2.5,
    label = L"\lfloor n/2 \rfloor",
    xlabel = L"n",
    ylabel = L"m",
    yticks = 0:5:N,
    xticks = 0:5:N,
    linestyle = :dot,
    # marker = :square,
    # markersize = 3,

    # Font sizes
    guidefontsize = 16,      # axis labels
    tickfontsize = 12,       # tick labels
    legendfontsize = 12,     # legend
    titlefontsize = 20,

    # Use Computer Modern (LaTeX-like) font
    fontfamily = "Computer Modern",
    legend = :topleft,
    foreground_color_legend = nothing,
        # ygrid = false
)

p=plot!(
    X, kte,
    lw = 2,              # line width
    marker = :diamond,    # symbol type
    # markersize = 2,
    label="m-TE",
    xlabel=L"n", ylabel=L"m",  yticks = 0:2:N
)
plot!(
    X, kuni,
    lw = 2,              # line width
    marker = :circle,    # symbol type
    # markersize = 2,
    label="m-uni",
    xlabel=L"n", ylabel=L"m",  yticks = 0:2:N
)

savefig(p, "k-TE.pdf")
```
