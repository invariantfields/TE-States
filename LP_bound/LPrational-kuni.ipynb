```python
# N number of qubits 

N = 40

kuni=[]

kend=0

for n in range(8,N+1):

    for k in range(floor(n/2), 0, -1):
       
        A,b,c=Standard_form(n,k, kuni=True) # constraints for k-uniform states
    
        dual = MixedIntegerLinearProgram(
        maximization=True,
        base_ring=QQ,
        solver="PPL"
        )
    
        y = dual.new_variable(real=True) # dual variables
        
        dual.set_objective(
            sum(b[i,0]*y[i] for i in  range(A.nrows()))
        ) # the objective function of the dual is b*y
        
        for j in range(n+1):
            dual.add_constraint(
                sum(A[i,j]*y[i] for i in range(A.nrows())) <= c[0,j]
            ) # y*A =< c is the dual contraints
        
        for j in range(A.nrows()):
            dual.add_constraint(y[j] >= 0) # y>=0 are positive


        dual.add_constraint(
            sum(b[i,0]*y[i] for i in  range(A.nrows())) <=1
        ) # Bounding the objective function of the dual. If y*b>=0, the result is infeasible
        
        dual.solve() # solve dual

        y_sol = matrix(QQ, [[dual.get_values(y[i]) for i in range(A.nrows())]]) # dual solution

        if y_sol*b==0:
            kuni.append(k)
            break

        """
        This condition checks if both solutions satisfy the respective conditions + strong duality, i.e.
    
        Primal            Dual
        
        min c*x^T           max b*y
        s.t. Ax >= b      s.t. yB <= c
             x_i >= 0            y_i >= 0
    
        Strong duality: c*x^T = y*b
        Weak duality: c*x^T - y*b >= 0

        since c=0, If y*b==0 if the solution is feasible and y*b>=0 otherwise.  
        
        """
```
